# Escalado
Conjunto de datos: iris (incluido en sklearn)
Este conjunto de datos es compacto y perfecto para demostrar técnicas de escalado.

```python
# Importar librerías necesarias
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Cargar el conjunto de datos
iris = load_iris(as_frame=True)
data = iris.data
```

```python
# Visualizar las estadísticas iniciales
print("Estadísticas iniciales:")
data.describe()
```

```python
# Aplicar MinMaxScaler
minmax_scaler = MinMaxScaler()
data_minmax = pd.DataFrame(minmax_scaler.fit_transform(data), columns=data.columns)
```

```python
# Aplicar StandardScaler
standard_scaler = StandardScaler()
data_standard = pd.DataFrame(standard_scaler.fit_transform(data), columns=data.columns)
```

### 2. Crea una gráfica de la distribución de la variable 'sepal length (cm)' antes y después del escalado.

```python
# Distribución de sepal length antes y después
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(data['sepal length (cm)'], bins=20, color='blue', alpha=0.7)
axes[0].set_title('Sepal Length Original')
axes[1].hist(data_minmax['sepal length (cm)'], bins=20, color='green', alpha=0.7)
axes[1].set_title('Sepal Length MinMax')
plt.tight_layout()
plt.show()
```

### 3. Crea una gráfica de dispersión entre 'sepal length (cm)' y 'sepal width (cm)' con los datos escalados por StandardScaler.

```python
# Dispersión con StandardScaler
plt.figure(figsize=(8, 6))
plt.scatter(data_standard['sepal length (cm)'], data_standard['sepal width (cm)'], alpha=0.7)
plt.xlabel('Sepal Length (Escalado)')
plt.ylabel('Sepal Width (Escalado)')
plt.title('Dispersión - StandardScaler')
plt.grid(True)
plt.show()
```

### 4. ¿Qué técnica de escalado crees que es más adecuada para este conjunto de datos? ¿Por qué?

```python
# Análisis comparativo
print("StandardScaler es mejor para este conjunto porque:")
print("- Los datos tienen distribuciones normales")
print("- StandardScaler centra en 0 y escala por desviación estándar")
print("- Preserva relaciones entre variables mejor que MinMax")
print(f"\nRango MinMax: [{data_minmax.min().min():.2f}, {data_minmax.max().max():.2f}]")
print(f"Media StandardScaler: {data_standard.mean().mean():.4f}")
```
