# Datos Faltantes

```python
# Importar librerías necesarias
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.impute import SimpleImputer, KNNImputer
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Cargar el conjunto de datos
diabetes = load_diabetes(as_frame=True)
data = diabetes.data
data['target'] = diabetes.target
```

```python
data.sample(10)
```

```python
# Simular valores faltantes
np.random.seed(42)
data.loc[data.sample(frac=0.1).index, 'bmi'] = np.nan
data.loc[data.sample(frac=0.1).index, 'bp'] = np.nan
```

### 1. ¿Cuántos valores faltantes hay en cada columna?**

```python
# Contar valores faltantes
print(data.isnull().sum())
```

### 2. Utiliza imputación simple (media) para llenar los valores faltantes de la columna 'bmi'.

```python
# Imputación simple con media
data_imputed_mean = data.copy()
imputer_mean = SimpleImputer(strategy='mean')
data_imputed_mean['bmi'] = imputer_mean.fit_transform(data[['bmi']])
print(f"Imputación completada. Valores faltantes: {data_imputed_mean['bmi'].isnull().sum()}")
```

### 3. Utiliza KNNImputer para imputar valores en las columnas 'bmi' y 'bp'. Compara los resultados con los de la imputación simple.

```python
# Imputación con KNNImputer
data_imputed_knn = data.copy()
knn_imputer = KNNImputer(n_neighbors=5)
data_imputed_knn[['bmi', 'bp']] = knn_imputer.fit_transform(data[['bmi', 'bp']])
print(f"Media 'bmi' - Original: {data['bmi'].mean():.4f}, Simple: {data_imputed_mean['bmi'].mean():.4f}, KNN: {data_imputed_knn['bmi'].mean():.4f}")
```

### 4. Genera un histograma comparando los datos antes y después de la imputación en la columna 'bmi'.

```python
# Histograma comparativo
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
axes[0].hist(data['bmi'].dropna(), bins=20, color='blue', alpha=0.7)
axes[0].set_title('Original')
axes[1].hist(data_imputed_mean['bmi'], bins=20, color='green', alpha=0.7)
axes[1].set_title('Simple')
axes[2].hist(data_imputed_knn['bmi'], bins=20, color='orange', alpha=0.7)
axes[2].set_title('KNN')
plt.tight_layout()
plt.show()
```
